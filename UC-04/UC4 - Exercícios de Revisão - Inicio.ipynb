## **Exercício 20** - Cadastro de produtos

**Objetivo:**

Aplicar os conteúdos aprendidos.

**Descrição:**

Crie um sistema simples de cadastro de produtos.

O sistema deve:

- pedir nome do produto;
- pedir preço;
- armazenar em um dicionário;
- adicionar em uma lista;
- perguntar se deseja continuar.

Ao final:

- mostrar todos os produtos cadastrados.


```python
produtos = {}

while True:
    produto = input('Digite o nome do produto: ')
    preco = float(input('Digite o preço do produto: '))

    produtos[produto] = preco

    x = input('Deseja continuar? (S/N)').lower()
    if x == 's':
        continue
    else:
        break

for produto, preco in produtos.items():
    print(f'Produto: {produto}, Preço: {preco}')

```
